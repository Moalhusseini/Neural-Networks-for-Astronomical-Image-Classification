# Neural Networks for Astronomical Image Classification: A Comparative Study of Quasars, Stars, and Galaxies Using SDSS Data

# Introduction and Background

Scientific research increasingly uses data science and artificial intelligence to understand complex problems that are difficult to solve through traditional methods alone. From studying the behavior of physical systems to analyzing medical images and predicting environmental changes, machine learning helps researchers identify patterns, test ideas, and extract useful information from large and complex datasets.

**Neural networks** are an important part of this development. They learn relationships between input data and target outputs through repeated training. Instead of requiring researchers to define every important pattern manually, neural networks can learn useful representations directly from data. This ability makes them valuable in scientific fields where the underlying patterns are complex, nonlinear, or difficult to describe with simple mathematical rules. However, their performance still depends on data quality, model design, and appropriate evaluation.

One important application of neural networks is **computer vision**, which focuses on enabling computers to interpret visual information. Image analysis includes several related tasks, such as image classification, object detection, image segmentation, and feature extraction. These techniques help computers recognize objects, distinguish between visual patterns, and identify important structures within images. Traditional image-processing methods often use manually designed features, while deep learning models, especially convolutional neural networks (CNNs), learn spatial patterns directly from image pixels.

Computer vision also provides useful tools for scientific research. In medicine, researchers use image analysis to study tissues and detect possible abnormalities. In environmental science, satellite images support land-cover mapping and the monitoring of changes on Earth. In materials science, image-based methods help identify structures and defects. In astronomy, computer vision supports the analysis of celestial objects, the study of galaxies, and the organization of the large amounts of visual data collected by modern surveys.

Astronomy presents a particularly interesting challenge because telescopes observe an enormous variety of objects across different distances, brightness levels, and physical conditions. Astronomical images contain information about the appearance, brightness distribution, and spatial structure of these objects. However, different objects may appear similar in an image, and the visual information alone does not always provide a reliable identification. Combining image analysis with trustworthy astronomical catalog information therefore creates an opportunity to investigate how effectively machine learning models recognize different classes of celestial objects.

This project focuses on three important classes: **quasars, stars, and galaxies**. Stars are astronomical objects that produce energy through processes in their interiors. Galaxies contain stars, gas, dust, and other components held together by gravity. Quasars are extremely luminous active galactic nuclei powered by matter accreting onto supermassive black holes. Although these objects have different physical properties, their observed images can vary in appearance, making their classification an interesting task for computer vision.

To investigate this problem, the project uses real astronomical data from the **Sloan Digital Sky Survey (SDSS)**, which provides extensive imaging and catalog information about the sky. The project connects this observational data with two machine learning approaches. First, a traditional machine learning model uses numerical features extracted from images, such as color information, intensity statistics, and spatial brightness patterns. Second, a convolutional neural network learns visual features directly from image pixels. Comparing these approaches helps examine the difference between manually designed image features and automatically learned representations.

The main purpose is not simply to train a model with high accuracy. Instead, the project investigates how different approaches perform on the same classification problem, evaluates their strengths and limitations, and considers what the results mean for astronomical image analysis. By combining astrophysics, image processing, computer vision, and machine learning, the project demonstrates how data-driven methods can support the analysis of scientific observations while maintaining the importance of reliable labels, fair evaluation, and careful interpretation.

## Methodology

The study follows a supervised image-classification workflow using data from the Sloan Digital Sky Survey (SDSS). The workflow combines catalog information, image preprocessing, traditional machine learning, and deep learning to classify astronomical objects.

First, SDSS catalog records are selected using their spectroscopic class labels. The catalog provides the object class and sky coordinates, while the corresponding image manifest identifies the retrieved images. The dataset contains 5,100 images distributed equally across three classes: GALAXY, QSO, and STAR.

Second, the catalog and image records are validated, and the images are prepared for analysis. The images are represented in RGB format and resized to a consistent resolution. Exploratory analysis is used to examine the selected sample and its visual and statistical properties.

Third, two classification approaches are developed:

1. **Traditional Machine Learning:** Numerical features are extracted from each RGB image. These include the mean and standard deviation of each color channel and normalized color histograms. Random Forest and XGBoost classifiers are trained using these extracted features.
2. **Convolutional Neural Network (CNN):** A CNN learns visual representations directly from RGB image pixels resized to 96 × 96 pixels and predicts one of the three astronomical classes.

The dataset is divided into development and test sets using stratified sampling, with 80% allocated to development and 20% reserved for testing. Model selection and cross-validation are conducted using the development data, while the test set is used for final performance evaluation.

Finally, the models are compared using accuracy, macro-averaged F1-score, macro ROC-AUC, class-level precision and recall, and confusion matrices. This comparison evaluates the predictive performance of the different approaches under the experimental setup used in this study.

## Inspecting the SDSS Data Access Service

Checks access to the official SDSS SkyServer service before retrieving catalog records. A small test query verifies that the service returns the expected object classifications and coordinates. These fields help identify suitable objects and locate their corresponding astronomical images.

```python
SDSS_SQL_URL = (
    "https://skyserver.sdss.org/dr17/"
    "SkyServerWS/SearchTools/SqlSearch"
)

query = """
SELECT TOP 10
    s.specObjID,
    s.class,
    s.z,
    p.objID,
    p.ra,
    p.dec
FROM SpecObj AS s
JOIN PhotoObj AS p
    ON s.bestObjID = p.objID
WHERE s.class IN ('QSO', 'STAR', 'GALAXY')
    AND s.zWarning = 0
ORDER BY s.specObjID
"""

response = requests.get(
    SDSS_SQL_URL,
    params={"cmd": query, "format": "csv"},
    timeout=60
)

print("HTTP status code:", response.status_code)
response.raise_for_status()

sdss_test = pd.read_csv(StringIO(response.text))

print("Returned data shape:", sdss_test.shape)
print("Returned columns:", sdss_test.columns.tolist())

display(sdss_test.head(10))
```

## Selecting a Balanced Catalog Sample

Retrieves a small test sample for each target class separately. Separate queries help verify the availability of quasars, stars, and galaxies before the project defines the final dataset size. The sample is used to inspect catalog availability, not to train the models.

```python
classes = ["QSO", "STAR", "GALAXY"]
class_samples = []

for object_class in classes:
    query = f"""
    SELECT TOP 10
        s.specObjID,
        s.class,
        s.z,
        s.zWarning,
        p.objID,
        p.ra,
        p.dec
    FROM SpecObj AS s
    JOIN PhotoObj AS p
        ON s.bestObjID = p.objID
    WHERE s.class = '{object_class}'
        AND s.zWarning = 0
    ORDER BY s.specObjID
    """

    response_class = requests.get(
        SDSS_SQL_URL,
        params={"cmd": query, "format": "csv"},
        timeout=60
    )

    response_class.raise_for_status()

    sample = pd.read_csv(
        StringIO(response_class.text),
        skiprows=1
    )

    class_samples.append(sample)

    print(f"{object_class}: {len(sample)} records returned")

sdss_class_test = pd.concat(
    class_samples,
    ignore_index=True
)

print("\nCombined dataset shape:", sdss_class_test.shape)
print("\nClass distribution:")
print(sdss_class_test["class"].value_counts())

display(sdss_class_test.head(10))
```

## Batch Image Retrieval and Quality Check

Retrieves images for the 30 catalog records collected during the initial test. It checks the response status, image format, and dimensions, then summarizes the retrieval results by class. This initial batch helps estimate image availability before the dataset is expanded.

```python
image_records = []
failed_images = []

for _, row in sdss_class_test.drop_duplicates(
    subset="objID"
).iterrows():

    params = {
        "ra": float(row["ra"]),
        "dec": float(row["dec"]),
        "scale": 0.396,
        "width": 128,
        "height": 128
    }

    try:
        img_response = requests.get(
            IMAGE_URL,
            params=params,
            timeout=60
        )

        img_response.raise_for_status()

        if not img_response.headers.get(
            "Content-Type", ""
        ).startswith("image/"):
            raise ValueError("Response is not an image.")

        img = Image.open(
            BytesIO(img_response.content)
        ).convert("RGB")

        if img.size != (128, 128):
            raise ValueError(
                f"Unexpected image dimensions: {img.size}"
            )

        image_records.append({
            "objID": row["objID"],
            "class": row["class"],
            "ra": row["ra"],
            "dec": row["dec"],
            "image": np.array(img)
        })

    except Exception as error:
        failed_images.append({
            "objID": row["objID"],
            "class": row["class"],
            "error": str(error)
        })

images_df = pd.DataFrame([
    {key: value for key, value in record.items()
     if key != "image"}
    for record in image_records
])

print("Unique catalog objects:", len(images_df) + len(failed_images))
print("Successfully retrieved images:", len(image_records))
print("Failed images:", len(failed_images))

print("\nSuccessful images by class:")
if not images_df.empty:
    print(images_df["class"].value_counts())

print("\nFailed images by class:")
if failed_images:
    display(pd.DataFrame(failed_images))
else:
    print("No image retrieval failures.")
```

## Downloading Images with Progress Tracking

Downloads SDSS images to Google Drive and displays a progress bar showing the completion percentage, processed images, and estimated remaining time. The manifest is updated regularly to track successful downloads and failures, allowing the process to resume more easily after an interruption.

```python
import time
from tqdm.auto import tqdm


BATCH_SIZE = 25
REQUEST_TIMEOUT = 30
PAUSE_SECONDS = 0.1

manifest_df = pd.read_csv(MANIFEST_PATH)

total_images = len(manifest_df)

completed_before = 0

for _, row in manifest_df.iterrows():
    image_path = Path(row["image_path"])

    if image_path.exists():
        try:
            with Image.open(image_path) as existing_image:
                existing_image.verify()

            completed_before += 1

        except Exception:
            image_path.unlink(missing_ok=True)

with tqdm(
    total=total_images,
    initial=completed_before,
    desc="Downloading SDSS images",
    unit="image",
    dynamic_ncols=True,
    smoothing=0.1
) as progress:

    for index, row in manifest_df.iterrows():

        image_path = Path(row["image_path"])
        image_path.parent.mkdir(parents=True, exist_ok=True)

        if image_path.exists():
            try:
                with Image.open(image_path) as existing_image:
                    existing_image.verify()

                manifest_df.at[index, "status"] = "success"
                manifest_df.at[index, "error"] = ""

                continue

            except Exception:
                image_path.unlink(missing_ok=True)

        try:
            params = {
                "ra": float(row["ra"]),
                "dec": float(row["dec"]),
                "scale": 0.396,
                "width": IMAGE_WIDTH,
                "height": IMAGE_HEIGHT
            }

            response_image = requests.get(
                IMAGE_URL,
                params=params,
                timeout=REQUEST_TIMEOUT
            )
            response_image.raise_for_status()

            if not response_image.headers.get(
                "Content-Type", ""
            ).startswith("image/"):
                raise ValueError("Response is not an image.")

            with Image.open(
                BytesIO(response_image.content)
            ) as downloaded_image:
                image = downloaded_image.convert("RGB")

            if image.size != (IMAGE_WIDTH, IMAGE_HEIGHT):
                raise ValueError(
                    f"Unexpected image size: {image.size}"
                )

            image.save(
                image_path,
                format="JPEG",
                quality=95
            )

            manifest_df.at[index, "status"] = "success"
            manifest_df.at[index, "error"] = ""

        except Exception as error:
            manifest_df.at[index, "status"] = "failed"
            manifest_df.at[index, "error"] = str(error)

        finally:
            progress.update(1)

            progress.set_postfix(
                success=int(
                    (manifest_df["status"] == "success").sum()
                ),
                failed=int(
                    (manifest_df["status"] == "failed").sum()
                )
            )

        if (index + 1) % BATCH_SIZE == 0:
            manifest_df.to_csv(
                MANIFEST_PATH,
                index=False
            )

        time.sleep(PAUSE_SECONDS)

manifest_df.to_csv(MANIFEST_PATH, index=False)

print("\nDownload process completed.")
print("\nOverall status:")
print(manifest_df["status"].value_counts())

print("\nStatus by class:")
display(
    pd.crosstab(
        manifest_df["class"],
        manifest_df["status"]
    )
)

print("\nManifest:", MANIFEST_PATH)
```

## Validate the Downloaded Image Dataset

```python
manifest_df = pd.read_csv(
    MANIFEST_PATH,
    dtype={
        "class": str,
        "status": str,
        "error": str
    }
)

validation_results = []

for _, row in tqdm(
    manifest_df.iterrows(),
    total=len(manifest_df),
    desc="Validating images",
    unit="image"
):
    image_path = Path(row["image_path"])

    result = {
        "objID": row["objID"],
        "class": row["class"],
        "image_path": str(image_path),
        "image_valid": False,
        "validation_error": ""
    }

    try:
        if not image_path.exists():
            raise FileNotFoundError("Image file does not exist.")

        with Image.open(image_path) as img:
            img.verify()

        with Image.open(image_path) as img:
            if img.size != (IMAGE_WIDTH, IMAGE_HEIGHT):
                raise ValueError(
                    f"Unexpected dimensions: {img.size}"
                )

            if img.mode not in ("RGB", "RGBA"):
                raise ValueError(
                    f"Unexpected image mode: {img.mode}"
                )

        result["image_valid"] = True

    except Exception as error:
        result["validation_error"] = str(error)

    validation_results.append(result)

validation_df = pd.DataFrame(validation_results)

print("Total records checked:", len(validation_df))
print("Valid images:", int(validation_df["image_valid"].sum()))
print("Invalid images:", int((~validation_df["image_valid"]).sum()))

print("\nValidation results by class:")
display(
    validation_df.groupby("class")["image_valid"]
    .agg(["sum", "count"])
    .rename(columns={"sum": "valid_images", "count": "total_records"})
)

VALIDATION_PATH = MANIFEST_DIR / "sdss_image_validation.csv"
validation_df.to_csv(VALIDATION_PATH, index=False)

print("\nValidation report saved to:", VALIDATION_PATH)

if (~validation_df["image_valid"]).any():
    display(validation_df.loc[~validation_df["image_valid"]])
```

## Retrying Failed Image Downloads

Retries the retrieval of images that are missing or invalid. It updates the manifest after each attempt and preserves the existing valid images. This helps complete the dataset without repeating the full download process.

```python
manifest_df = pd.read_csv(
    MANIFEST_PATH,
    dtype={
        "class": str,
        "status": str,
        "error": str
    }
)

invalid_ids = set(
    validation_df.loc[
        ~validation_df["image_valid"], "objID"
    ].astype(str)
)

retry_indices = manifest_df.index[
    manifest_df["objID"].astype(str).isin(invalid_ids)
].tolist()

print("Images requiring retry:", len(retry_indices))

for index in tqdm(retry_indices, desc="Retrying images"):
    row = manifest_df.loc[index]
    image_path = Path(row["image_path"])

    try:
        params = {
            "ra": float(row["ra"]),
            "dec": float(row["dec"]),
            "scale": 0.396,
            "width": IMAGE_WIDTH,
            "height": IMAGE_HEIGHT
        }

        response_image = requests.get(
            IMAGE_URL,
            params=params,
            timeout=60
        )
        response_image.raise_for_status()

        if not response_image.headers.get(
            "Content-Type", ""
        ).startswith("image/"):
            raise ValueError("Response is not an image.")

        with Image.open(
            BytesIO(response_image.content)
        ) as downloaded_image:
            image = downloaded_image.convert("RGB")

        if image.size != (IMAGE_WIDTH, IMAGE_HEIGHT):
            raise ValueError(
                f"Unexpected dimensions: {image.size}"
            )

        image.save(image_path, format="JPEG", quality=95)

        manifest_df.at[index, "status"] = "success"
        manifest_df.at[index, "error"] = ""

    except Exception as error:
        manifest_df.at[index, "status"] = "failed"
        manifest_df.at[index, "error"] = str(error)

    manifest_df.to_csv(MANIFEST_PATH, index=False)
    time.sleep(0.2)

print("\nUpdated download status:")
print(manifest_df["status"].value_counts())
```

## Final Image Validation

Repeats the image validation after the retry. It confirms that all image files exist, can be opened, and have the expected dimensions. The final dataset summary records the number of valid images in each class before preprocessing begins.

```python
manifest_df = pd.read_csv(
    MANIFEST_PATH,
    dtype={
        "class": str,
        "status": str,
        "error": str
    }
)

final_validation = []

for _, row in tqdm(
    manifest_df.iterrows(),
    total=len(manifest_df),
    desc="Final validation",
    unit="image"
):
    image_path = Path(row["image_path"])

    valid = False
    error_message = ""

    try:
        if not image_path.exists():
            raise FileNotFoundError("Image file does not exist.")

        with Image.open(image_path) as img:
            img.verify()

        with Image.open(image_path) as img:
            if img.size != (IMAGE_WIDTH, IMAGE_HEIGHT):
                raise ValueError(f"Unexpected dimensions: {img.size}")

            if img.mode != "RGB":
                raise ValueError(f"Unexpected mode: {img.mode}")

        valid = True

    except Exception as error:
        error_message = str(error)

    final_validation.append({
        "objID": row["objID"],
        "class": row["class"],
        "image_path": str(image_path),
        "image_valid": valid,
        "validation_error": error_message
    })

final_validation_df = pd.DataFrame(final_validation)

FINAL_VALIDATION_PATH = (
    MANIFEST_DIR / "sdss_final_image_validation.csv"
)

final_validation_df.to_csv(
    FINAL_VALIDATION_PATH,
    index=False
)

print("Total records:", len(final_validation_df))
print("Valid images:", int(final_validation_df["image_valid"].sum()))
print("Invalid images:", int((~final_validation_df["image_valid"]).sum()))

print("\nFinal valid image counts:")
display(
    final_validation_df.groupby("class")["image_valid"]
    .agg(["sum", "count"])
    .rename(columns={"sum": "valid_images", "count": "total_records"})
)

print("\nValidation report saved to:", FINAL_VALIDATION_PATH)
```

## RGB Feature Extraction Pilot Test

Extracts 30 color-based features from one image to verify the implementation before processing the full development dataset.

```python
def extract_rgb_features(image_path):
    with Image.open(image_path) as image:
        image = image.convert("RGB").resize((128, 128))
        rgb = np.asarray(image, dtype=np.float32) / 255.0

    features = []
    feature_names = []

    channel_names = ["R", "G", "B"]

    # 1. Mean and standard deviation for each RGB channel
    for channel_index, channel_name in enumerate(channel_names):
        channel = rgb[:, :, channel_index]

        features.extend([
            channel.mean(),
            channel.std()
        ])

        feature_names.extend([
            f"{channel_name}_mean",
            f"{channel_name}_std"
        ])

    # 2. Eight-bin normalized histogram for each RGB channel
    for channel_index, channel_name in enumerate(channel_names):
        histogram, _ = np.histogram(
            rgb[:, :, channel_index],
            bins=8,
            range=(0.0, 1.0)
        )

        histogram = histogram.astype(np.float32)
        histogram /= histogram.sum()

        features.extend(histogram.tolist())

        feature_names.extend([
            f"{channel_name}_hist_{i}"
            for i in range(8)
        ])

    feature_vector = np.asarray(features, dtype=np.float32)

    assert len(feature_vector) == 30
    assert np.isfinite(feature_vector).all()

    return feature_vector, feature_names


# Test on one image from the development set
pilot_path = development_df.iloc[0]["image_path"]

pilot_features, pilot_feature_names = extract_rgb_features(
    pilot_path
)

pilot_df = pd.DataFrame({
    "feature": pilot_feature_names,
    "value": pilot_features
})

print("Number of features:", len(pilot_features))
print("All values finite:", np.isfinite(pilot_features).all())

display(pilot_df)
```

## Deep Learning: CNN-Based Image Classification

A Convolutional Neural Network (CNN) is used to classify astronomical images directly from their visual information. Unlike Random Forest and XGBoost, which rely on extracted statistical features, the CNN learns spatial patterns from the image pixels during training.

### Prepare the Image Dataset

The original RGB images are resized to 96 × 96 pixels, and their pixel values are normalized to the range [0, 1]. The existing development and independent test split is preserved. A separate validation subset is used during training to monitor model performance and select the best model. The test images are not used for training or validation.

### Import The Deep Learning Library & Verificate The Training Enviroment

```python
import tensorflow as tf

print("TensorFlow version:", tf.__version__)
print("Available GPUs:", tf.config.list_physical_devices("GPU"))

CLASS_NAMES = ["GALAXY", "QSO", "STAR"]

CLASS_TO_INDEX = {
    class_name: index
    for index, class_name in enumerate(CLASS_NAMES)
}

print("Class mapping:", CLASS_TO_INDEX)
```

### Load and Preprocess Images

The internal validation subset is used to monitor training and select the best epoch. Stratification preserves the class distribution.

```python
cnn_train_df, cnn_val_df = train_test_split(
    development_df,
    test_size=0.10,
    random_state=42,
    stratify=development_df["class"]
)

print("CNN training images:", len(cnn_train_df))
print("CNN validation images:", len(cnn_val_df))
print("Independent test images:", len(test_df))

print("\nTraining distribution:")
print(cnn_train_df["class"].value_counts().reindex(CLASS_NAMES))

print("\nValidation distribution:")
print(cnn_val_df["class"].value_counts().reindex(CLASS_NAMES))
```

### Create Image Arrays

Images are loaded as RGB arrays and normalized by dividing pixel values by 255. Labels are converted into integer class indices for the CNN output layer.

```python
def load_image_array(image_path):
    with Image.open(image_path) as image:
        image = image.convert("RGB").resize((128, 128))
        array = np.asarray(image, dtype=np.float32) / 255.0
    return array


def prepare_cnn_data(dataframe):
    images = np.stack([
        load_image_array(path)
        for path in tqdm(
            dataframe["image_path"],
            desc="Loading images"
        )
    ])

    labels = np.array([
        CLASS_TO_INDEX[label]
        for label in dataframe["class"]
    ], dtype=np.int32)

    return images, labels


X_cnn_train, y_cnn_train = prepare_cnn_data(cnn_train_df)
X_cnn_val, y_cnn_val = prepare_cnn_data(cnn_val_df)

print("Training image tensor:", X_cnn_train.shape)
print("Training labels:", y_cnn_train.shape)

print("Validation image tensor:", X_cnn_val.shape)
print("Validation labels:", y_cnn_val.shape)

assert X_cnn_train.shape[1:] == (128, 128, 3)
assert np.isfinite(X_cnn_train).all()
assert np.isfinite(X_cnn_val).all()
```

### Resize Images to 96×96

All training and validation images are resized to 96×96 pixels. The original class labels and dataset split remain unchanged.

```python
IMAGE_SIZE = 96

X_cnn_train_96 = tf.image.resize(
    X_cnn_train,
    (IMAGE_SIZE, IMAGE_SIZE)
).numpy().astype(np.float32)

X_cnn_val_96 = tf.image.resize(
    X_cnn_val,
    (IMAGE_SIZE, IMAGE_SIZE)
).numpy().astype(np.float32)

print("Training images:", X_cnn_train_96.shape)
print("Validation images:", X_cnn_val_96.shape)
print("Training labels:", y_cnn_train.shape)
print("Validation labels:", y_cnn_val.shape)

assert X_cnn_train_96.shape[1:] == (96, 96, 3)
assert X_cnn_val_96.shape[1:] == (96, 96, 3)
assert np.isfinite(X_cnn_train_96).all()
assert np.isfinite(X_cnn_val_96).all()
```

### Build the CNN Architecture

The CNN learns visual features directly from the RGB images. Dropout is included to reduce overfitting, and the output layer predicts the three astronomical classes.

```python
from tensorflow.keras import layers, models

tf.keras.utils.set_random_seed(42)

cnn_model = models.Sequential([
    layers.Input(shape=(96, 96, 3)),

    layers.Conv2D(32, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(64, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    layers.Conv2D(128, (3, 3), activation="relu", padding="same"),
    layers.MaxPooling2D((2, 2)),

    layers.GlobalAveragePooling2D(),

    layers.Dense(128, activation="relu"),
    layers.Dropout(0.40),

    layers.Dense(3, activation="softmax")
])

cnn_model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

cnn_model.summary()
```

### Train the CNN

The CNN is trained using the training subset. Validation loss is monitored to retain the best model and reduce unnecessary training. The independent test set remains unused.

```python
MODEL_DIR = PROJECT_DIR / "models"
MODEL_DIR.mkdir(parents=True, exist_ok=True)

BEST_MODEL_PATH_96 = MODEL_DIR / "cnn_96px_best.keras"

callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=5,
        restore_best_weights=True,
        verbose=1
    ),
    tf.keras.callbacks.ModelCheckpoint(
        filepath=str(BEST_MODEL_PATH_96),
        monitor="val_loss",
        save_best_only=True,
        verbose=1
    )
]

history_96 = cnn_model.fit(
    X_cnn_train_96,
    y_cnn_train,
    validation_data=(X_cnn_val_96, y_cnn_val),
    epochs=25,
    batch_size=32,
    callbacks=callbacks,
    shuffle=True,
    verbose=1
)

print("Training completed.")
print("Best model:", BEST_MODEL_PATH_96)
print("Epochs completed:", len(history_96.history["loss"]))
```

### CNN: Independent Test Evaluation

In this step, the trained Convolutional Neural Network (CNN) is evaluated on the independent test set using the original RGB images resized to 96 × 96 pixels. Classification metrics, a confusion matrix, and multiclass ROC curves are used to assess its performance across the three astronomical classes. ## RGB-Based Feature Extraction

### 1. Overview

In addition to the Convolutional Neural Network (CNN), this study uses traditional machine learning models to classify astronomical objects into three classes: Quasars (QSO), Stars (STAR), and Galaxies (GALAXY). Since traditional machine learning algorithms require numerical input features, the RGB images were converted into a set of statistical color features.

The feature extraction process was designed to provide a simple representation of the color information contained in each image. A total of 30 features were extracted from every image using the mean, standard deviation, and normalized histograms of the red, green, and blue channels.

### 2. Image Preparation

Before extracting the features, each image was converted to RGB format and resized to (128 times 128) pixels. The image was then converted into a NumPy array with floating-point values normalized to the range ([0,1]).

Each image contains three color channels: red (R), green (G), and blue (B). Each channel contains 16,384 pixel values after resizing. This consistent image representation ensures that the same feature extraction procedure can be applied to all samples.

### 3. Statistical Color Features

The first group consists of six statistical features: the mean and standard deviation of each RGB channel. These features summarize the average color intensity and its variation across the image.

#### 3.1. Mean Intensity

The mean intensity represents the average pixel value within a color channel. It provides a general measure of the channel's intensity across the entire image.

For a color channel (c), the mean is calculated as:



$$
mu_c = \frac{1}{N}\sum\_{i=1}^{N}x_i
$$



where (x_i) is the normalized intensity of pixel (i), and (N) is the total number of pixels in the channel.

Three mean features are extracted:

- `R_mean`: Mean red-channel intensity.
- `G_mean`: Mean green-channel intensity.
- `B_mean`: Mean blue-channel intensity.

#### 3.2. Standard Deviation

The standard deviation measures the variation in pixel intensity within each color channel. A higher value indicates greater variation in the channel's pixel intensities, while a lower value indicates that the intensities are more similar.

The standard deviation is calculated as:

$$
\sigma_c =
\sqrt{\frac{1}{N}\sum\_{i=1}^{N}(x_i-\mu_c)^2}
$$

The implementation uses NumPy's default standard deviation calculation, which divides by (N).

The extracted features are:

- `R_std`: Standard deviation of the red channel.
- `G_std`: Standard deviation of the green channel.
- `B_std`: Standard deviation of the blue channel.

Together, the mean and standard deviation features provide a compact summary of the overall color intensity and its variation in each image.

### 4. Normalized RGB Histograms

The second group consists of 24 features derived from color histograms. Unlike the mean and standard deviation, histograms describe how pixel intensities are distributed across different intensity ranges.

For each RGB channel, the normalized intensity range from 0 to 1 is divided into eight equal-width bins. The number of pixels falling within each bin is counted and divided by the total number of pixels.

The normalized histogram value for bin (k) is:

$$
h_k = \frac{n_k}{N}
$$

where (n_k) represents the number of pixels in bin (k), and (N) is the total number of pixels in the channel.

Each channel produces eight histogram features:

- `R_hist_0` to `R_hist_7`: Red-channel histogram.
- `G_hist_0` to `G_hist_7`: Green-channel histogram.
- `B_hist_0` to `B_hist_7`: Blue-channel histogram.

Since the histograms are normalized, the eight values for each channel sum to approximately 1, subject to floating-point precision.

This representation captures differences in the distribution of color intensities that may not be fully described by the mean and standard deviation alone.

### 5. Final Feature Representation

The extracted features are combined into a single numerical vector for each image. The final representation contains 30 features, as summarized in Table X.

| Feature Group           | Description                                  | Number of Features |
| ----------------------- | -------------------------------------------- | -----------------: |
| RGB means               | Mean intensity for each color channel        |                  3 |
| RGB standard deviations | Intensity variation for each color channel   |                  3 |
| RGB histograms          | Eight normalized bins for each color channel |                 24 |
| **Total**               | **Statistical RGB representation**           |             **30** |

The feature extraction function also verifies that the resulting vector contains exactly 30 values and that all values are finite. These checks help identify implementation errors before the features are used for model training.

### 6. Role in Traditional Machine Learning

The resulting feature vectors are used as inputs to the traditional machine learning models, including Random Forest and XGBoost. Both models learn relationships between the extracted color statistics and the known astronomical class labels.

This approach provides a relatively simple and computationally efficient baseline for comparison with the CNN, which learns image representations directly from pixel data.

The two approaches differ in how they represent the input. The traditional models rely on predefined statistical color features, whereas the CNN can learn more complex spatial and visual patterns from the images during training.

### 7. Limitations

The extracted features describe the statistical distribution of RGB pixel intensities rather than calibrated astronomical measurements. The images are RGB representations of SDSS observations, and the extracted values should therefore be interpreted as image-based color statistics.

Furthermore, these 30 features do not explicitly describe spatial structures such as galaxy arms, object morphology, or the arrangement of sources within an image. They may also be affected by background pixels, image noise, and the image rendering process.

Consequently, this feature set provides a simplified representation of the images rather than a complete physical description of the astronomical objects. Its usefulness is evaluated through the performance of the traditional machine learning models and their comparison with the CNN.

### 8. Summary

In this study, 30 statistical color features are extracted from each (128 \times 128) RGB image. These features consist of six summary statistics and 24 normalized histogram values. The resulting feature vectors provide the numerical input required by the traditional machine learning models and establish a baseline for comparison with the image-based CNN approach.


# Discussion

The results indicate that astronomical image classification can benefit from models that learn visual patterns directly from images. The CNN performed best on galaxies, while stars were the most challenging class, based on the recorded test results.

The confusion between stars and QSOs suggests that distinguishing these classes was difficult for the model using the available image representation. However, further analysis would be required to determine which image characteristics contributed to these errors.

The training and validation curves showed no clear evidence of severe overfitting. Training accuracy reached 83.82%, while the best validation accuracy was 82.11%, with a best validation loss of 0.4926. Although these results are encouraging, they do not establish generalization to other astronomical surveys or observing conditions. Evaluation on an external dataset would be needed to assess broader applicability.

# Limitations

Several limitations should be considered when interpreting the results.

First, the dataset contained equal numbers of galaxies, QSOs, and stars. This balanced design supports class-wise comparison, but it does not reflect the natural distribution of astronomical objects in the sky.

Second, the images were RGB SDSS cutouts resized for model training. Their visual information depends on the image-rendering process, resolution, and selected cutout settings. The models therefore classify the available image representations rather than working directly with calibrated astronomical measurements.

Third, the evaluation used a held-out test set from the same dataset. Although this provides an independent assessment within the selected sample, it does not establish performance on other surveys, instruments, or observing conditions.

Finally, the study compared three models under the selected experimental settings. Different architectures, image resolutions, feature representations, or hyperparameter choices may produce different results.

# Future Work

Future work could extend this study in several directions. Testing the trained models on an external astronomical dataset would provide a stronger assessment of generalization. Additional experiments could also examine the effect of image resolution, data augmentation, and alternative CNN architectures on classification performance.

Another useful direction would be to combine image-based learning with selected catalog features, such as colors and redshift, to investigate whether these complementary sources of information improve classification. Such experiments should be designed carefully to prevent data leakage and ensure that the test set remains independent.

Finally, model interpretability methods could help identify which image regions contribute most to the CNN's predictions. This may provide a better understanding of the visual patterns associated with each class and help investigate the errors observed between stars and QSOs.

# Conclusion

This study compared Random Forest, XGBoost, and a convolutional neural network (CNN) for classifying SDSS astronomical images into galaxies, quasars, and stars. Among the evaluated models, the CNN achieved the strongest overall test performance, indicating that learning visual features directly from images was effective for this task.

The results also showed that classification performance varied across classes, with galaxies identified most successfully and stars presenting greater difficulty. The training and validation results were encouraging, although further evaluation on external datasets is necessary to establish broader generalization.

Overall, this project demonstrates the potential of deep learning for astronomical image classification and provides a foundation for future experiments involving richer astronomical features, external validation, and model interpretability.
